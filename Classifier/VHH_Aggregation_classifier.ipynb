# Decision Tree Classifier for Predicting Aggregation Propensity of VHHs

Objective: The goal of this notebook is to use a decision tree classifier to predict the aggregation propensity of VHHs.

```python
import pandas as pd
from sklearn.metrics import accuracy_score
from sklearn.metrics import balanced_accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
import joblib
```

```python
# import the data
df = pd.read_csv('Classifier_data.csv')
```

```python
# separate into X and y
df_X = df[['app_charge', 'hyd_moment']]
df_y = df['Main Label']
```

```python
# import the model
dtc = joblib.load('VHH_Classifier.pkl')
dtc
```

```python
# predict VHH aggregation
prediction = dtc.predict(df_X)

# evaluate
print('Accuracy', f"{accuracy_score(df_y,prediction):1.2f}")
print('Balanced Accuracy', f"{balanced_accuracy_score(df_y,prediction):1.2f}")
print('Precision', f"{precision_score(df_y,prediction):1.2f}")
print('Recall', f"{recall_score(df_y,prediction):1.2f}")
print('F1 Score', f"{f1_score(df_y,prediction):1.2f}")
```
